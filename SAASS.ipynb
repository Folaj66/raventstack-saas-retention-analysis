Exploratory Data Analysis

```python
# Phase 3 Context:
# This phase creates the analysis tables needed for churn analysis, activation analysis,
# cohort retention, monthly revenue, and NRR reporting.

accounts["signup_month"] = accounts["signup_date"].dt.to_period("M").dt.to_timestamp()
subscriptions["start_month"] = subscriptions["start_date"].dt.to_period("M").dt.to_timestamp()
subscriptions["end_month"] = subscriptions["effective_end_date"].dt.to_period("M").dt.to_timestamp()
feature_usage["usage_month"] = feature_usage["usage_date"].dt.to_period("M").dt.to_timestamp()
support_tickets["submitted_month"] = support_tickets["submitted_at"].dt.to_period("M").dt.to_timestamp()
churn_events["churn_month"] = churn_events["churn_date"].dt.to_period("M").dt.to_timestamp()
```

```python
# Churn by plan tier

plan_churn = (
    accounts
    .groupby("plan_tier")
    .agg(
        total_accounts=("account_id", "nunique"),
        churned_accounts=("churn_flag", "sum")
    )
    .reset_index()
)

plan_churn["churn_rate"] = plan_churn["churned_accounts"] / plan_churn["total_accounts"]

plan_churn.sort_values("churn_rate", ascending=False)
```

```python
# Churn by industry

industry_churn = (
    accounts
    .groupby("industry")
    .agg(
        total_accounts=("account_id", "nunique"),
        churned_accounts=("churn_flag", "sum")
    )
    .reset_index()
)

industry_churn["churn_rate"] = industry_churn["churned_accounts"] / industry_churn["total_accounts"]

industry_churn.sort_values("churn_rate", ascending=False)
```

```python
# Churn by country

country_churn = (
    accounts
    .groupby("country")
    .agg(
        total_accounts=("account_id", "nunique"),
        churned_accounts=("churn_flag", "sum")
    )
    .reset_index()
)

country_churn["churn_rate"] = country_churn["churned_accounts"] / country_churn["total_accounts"]

country_churn.sort_values("churn_rate", ascending=False)
```

```python
# Churn reasons

churn_reasons = (
    churn_events
    .groupby("reason_code")
    .agg(
        churn_events=("churn_event_id", "count"),
        affected_accounts=("account_id", "nunique"),
        avg_refund_amount=("refund_amount_usd", "mean"),
        total_refund_amount=("refund_amount_usd", "sum"),
        feedback_missing_rate=("feedback_missing", "mean")
    )
    .reset_index()
    .sort_values("churn_events", ascending=False)
)

churn_reasons
```

```python
# Support experience by account

support_account_summary = (
    support_tickets
    .groupby("account_id")
    .agg(
        ticket_count=("ticket_id", "count"),
        avg_resolution_time_hours=("resolution_time_hours", "mean"),
        avg_first_response_time_minutes=("first_response_time_minutes", "mean"),
        escalation_count=("escalation_flag", "sum"),
        avg_satisfaction_score=("satisfaction_score_imputed", "mean"),
        unrated_ticket_count=("satisfaction_missing", "sum")
    )
    .reset_index()
)

support_churn_analysis = (
    accounts[["account_id", "churn_flag", "plan_tier", "industry", "country"]]
    .merge(support_account_summary, on="account_id", how="left")
)

support_columns = [
    "ticket_count",
    "avg_resolution_time_hours",
    "avg_first_response_time_minutes",
    "escalation_count",
    "avg_satisfaction_score",
    "unrated_ticket_count"
]

support_churn_analysis[support_columns] = support_churn_analysis[support_columns].fillna(0)

support_churn_analysis.head()
```

```python
# Compare support experience for churned vs retained customers

support_churn_summary = (
    support_churn_analysis
    .groupby("churn_flag")
    .agg(
        accounts=("account_id", "nunique"),
        avg_ticket_count=("ticket_count", "mean"),
        avg_resolution_time_hours=("avg_resolution_time_hours", "mean"),
        avg_first_response_time_minutes=("avg_first_response_time_minutes", "mean"),
        avg_escalation_count=("escalation_count", "mean"),
        avg_satisfaction_score=("avg_satisfaction_score", "mean"),
        avg_unrated_ticket_count=("unrated_ticket_count", "mean")
    )
    .reset_index()
)

support_churn_summary
```

```python
# First 7 days product activation
# This identifies product usage during the first 7 days after subscription start.

activation_base = (
    feature_usage
    .merge(
        subscriptions[["subscription_id", "account_id", "start_date"]],
        on="subscription_id",
        how="left"
    )
)

activation_base["days_since_subscription_start"] = (
    activation_base["usage_date"] - activation_base["start_date"]
).dt.days

activation_7d = activation_base[
    (activation_base["days_since_subscription_start"] >= 0) &
    (activation_base["days_since_subscription_start"] <= 7)
].copy()

activation_7d.head()
```

```python
# Activation summary by account

activation_account_summary = (
    activation_7d
    .groupby("account_id")
    .agg(
        activation_events_7d=("usage_id", "count"),
        activation_features_used_7d=("feature_name", "nunique"),
        activation_total_usage_7d=("usage_count", "sum"),
        activation_total_duration_secs_7d=("usage_duration_secs", "sum"),
        activation_total_errors_7d=("error_count", "sum"),
        beta_feature_used_7d=("is_beta_feature", "max")
    )
    .reset_index()
)

activation_churn_analysis = (
    accounts[["account_id", "churn_flag", "plan_tier", "industry", "country"]]
    .merge(activation_account_summary, on="account_id", how="left")
)

activation_columns = [
    "activation_events_7d",
    "activation_features_used_7d",
    "activation_total_usage_7d",
    "activation_total_duration_secs_7d",
    "activation_total_errors_7d",
    "beta_feature_used_7d"
]

activation_churn_analysis[activation_columns] = activation_churn_analysis[activation_columns].fillna(0)

activation_churn_analysis.head()
```

```python
# Compare activation behavior for churned vs retained customers

activation_churn_summary = (
    activation_churn_analysis
    .groupby("churn_flag")
    .agg(
        accounts=("account_id", "nunique"),
        avg_activation_events_7d=("activation_events_7d", "mean"),
        avg_features_used_7d=("activation_features_used_7d", "mean"),
        avg_total_usage_7d=("activation_total_usage_7d", "mean"),
        avg_duration_secs_7d=("activation_total_duration_secs_7d", "mean"),
        avg_errors_7d=("activation_total_errors_7d", "mean"),
        beta_feature_usage_rate_7d=("beta_feature_used_7d", "mean")
    )
    .reset_index()
)

activation_churn_summary
```

```python
# Compare activation behavior for churned vs retained customers

activation_churn_summary = (
    activation_churn_analysis
    .groupby("churn_flag")
    .agg(
        accounts=("account_id", "nunique"),
        avg_activation_events_7d=("activation_events_7d", "mean"),
        avg_features_used_7d=("activation_features_used_7d", "mean"),
        avg_total_usage_7d=("activation_total_usage_7d", "mean"),
        avg_duration_secs_7d=("activation_total_duration_secs_7d", "mean"),
        avg_errors_7d=("activation_total_errors_7d", "mean"),
        beta_feature_usage_rate_7d=("beta_feature_used_7d", "mean")
    )
    .reset_index()
)

activation_churn_summary
```

```python
# Feature-level activation and churn analysis

feature_activation_retention = (
    activation_7d
    .merge(accounts[["account_id", "churn_flag"]], on="account_id", how="left")
    .groupby("feature_name")
    .agg(
        accounts_using_feature_7d=("account_id", "nunique"),
        churned_accounts=("churn_flag", "sum"),
        avg_usage_count=("usage_count", "mean"),
        avg_duration_secs=("usage_duration_secs", "mean"),
        avg_error_count=("error_count", "mean")
    )
    .reset_index()
)

feature_activation_retention["churn_rate"] = (
    feature_activation_retention["churned_accounts"] /
    feature_activation_retention["accounts_using_feature_7d"]
)

feature_activation_retention.sort_values("churn_rate")
```

```python
# Cohort retention analysis
# Cohort month is based on the subscription start month.

cohort_base = subscriptions.copy()
cohort_base["cohort_month"] = cohort_base["start_month"]

month_range = pd.date_range(
    cohort_base["start_month"].min(),
    analysis_date,
    freq="MS"
)

cohort_rows = []

for _, row in cohort_base.iterrows():
    active_months = month_range[
        (month_range >= row["start_month"]) &
        (month_range <= row["end_month"])
    ]

    for active_month in active_months:
        cohort_rows.append({
            "subscription_id": row["subscription_id"],
            "account_id": row["account_id"],
            "cohort_month": row["cohort_month"],
            "active_month": active_month,
            "mrr_amount": row["mrr_amount"],
            "arr_amount": row["arr_amount"],
            "plan_tier": row["plan_tier"],
            "subscription_status": row["subscription_status"]
        })

cohort_activity = pd.DataFrame(cohort_rows)

cohort_activity["cohort_index"] = (
    (cohort_activity["active_month"].dt.year - cohort_activity["cohort_month"].dt.year) * 12 +
    (cohort_activity["active_month"].dt.month - cohort_activity["cohort_month"].dt.month)
)

cohort_activity.head()
```

```python
# Cohort retention matrix

cohort_counts = (
    cohort_activity
    .groupby(["cohort_month", "cohort_index"])
    .agg(active_accounts=("account_id", "nunique"))
    .reset_index()
)

cohort_sizes = (
    cohort_counts[cohort_counts["cohort_index"] == 0]
    [["cohort_month", "active_accounts"]]
    .rename(columns={"active_accounts": "cohort_size"})
)

cohort_retention = cohort_counts.merge(cohort_sizes, on="cohort_month", how="left")

cohort_retention["retention_rate"] = (
    cohort_retention["active_accounts"] / cohort_retention["cohort_size"]
)

cohort_retention_matrix = cohort_retention.pivot_table(
    index="cohort_month",
    columns="cohort_index",
    values="retention_rate"
)

cohort_retention_matrix
```

```python
# Monthly revenue trend

monthly_revenue = (
    cohort_activity
    .groupby("active_month")
    .agg(
        active_accounts=("account_id", "nunique"),
        active_subscriptions=("subscription_id", "nunique"),
        total_mrr=("mrr_amount", "sum"),
        total_arr=("arr_amount", "sum")
    )
    .reset_index()
    .sort_values("active_month")
)

monthly_revenue["mrr_growth_rate"] = monthly_revenue["total_mrr"].pct_change()
monthly_revenue["arr_growth_rate"] = monthly_revenue["total_arr"].pct_change()

monthly_revenue.head()
```

```python
# Monthly churn analysis

monthly_churn = (
    churn_events
    .groupby("churn_month")
    .agg(
        churn_events=("churn_event_id", "count"),
        churned_accounts=("account_id", "nunique"),
        total_refunds=("refund_amount_usd", "sum")
    )
    .reset_index()
    .rename(columns={"churn_month": "month"})
)

monthly_active = (
    monthly_revenue
    .rename(columns={"active_month": "month"})
    [["month", "active_accounts", "total_mrr", "total_arr"]]
)

monthly_churn_analysis = monthly_active.merge(monthly_churn, on="month", how="left")

monthly_churn_analysis[["churn_events", "churned_accounts", "total_refunds"]] = (
    monthly_churn_analysis[["churn_events", "churned_accounts", "total_refunds"]].fillna(0)
)

monthly_churn_analysis["account_churn_rate"] = (
    monthly_churn_analysis["churned_accounts"] / monthly_churn_analysis["active_accounts"]
)

monthly_churn_analysis.head()
```

```python
# NRR analysis
# This is a simplified NRR table using monthly active MRR and churned MRR.

new_revenue = (
    subscriptions
    .groupby("start_month")
    .agg(
        new_mrr=("mrr_amount", "sum"),
        new_arr=("arr_amount", "sum")
    )
    .reset_index()
    .rename(columns={"start_month": "month"})
)

churned_revenue = (
    subscriptions[subscriptions["subscription_status"] == "Ended"]
    .groupby("end_month")
    .agg(
        churned_mrr=("mrr_amount", "sum"),
        churned_arr=("arr_amount", "sum")
    )
    .reset_index()
    .rename(columns={"end_month": "month"})
)

nrr_table = (
    monthly_active
    .merge(new_revenue, on="month", how="left")
    .merge(churned_revenue, on="month", how="left")
)

nrr_table[["new_mrr", "new_arr", "churned_mrr", "churned_arr"]] = (
    nrr_table[["new_mrr", "new_arr", "churned_mrr", "churned_arr"]].fillna(0)
)

nrr_table["starting_mrr"] = nrr_table["total_mrr"].shift(1)

nrr_table["gross_revenue_churn_rate"] = (
    nrr_table["churned_mrr"] / nrr_table["starting_mrr"]
)

nrr_table["nrr"] = (
    (nrr_table["starting_mrr"] - nrr_table["churned_mrr"]) /
    nrr_table["starting_mrr"]
)

nrr_table.head()
```
